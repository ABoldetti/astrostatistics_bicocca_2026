```python
import numpy as np
from sklearn.mixture import GaussianMixture
import matplotlib.pyplot as plt
import os
```

```python
a = np.load("../solutions/formationchannels.npy")

plt.hist(a , bins = 80);
```

```python
model = GaussianMixture(n_components=3)

model.fit(a)

model.weights_
plt.hist(a, bins = 80)
x = np.linspace( np.min(a) , np.max(a), 1000)
from scipy.stats import norm
for m , mu , sigma in zip( model.weights_ , model.means_, model.covariances_):
    plt.plot( x , m*norm.pdf(x, loc = mu[0] , scale = sigma[0][0]))



```
