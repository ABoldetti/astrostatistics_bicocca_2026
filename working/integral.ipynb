```python
import numpy as np
import scipy.stats as sp

import matplotlib.pyplot as plt
```

```python

tau = 10000
size = int(1e7)

initial_dist = sp.expon( 0 , tau)

func = lambda x: x**3 * sp.norm.pdf(x, 0, 1) / sp.expon.pdf(x, loc=0, scale=tau)/tau

data = initial_dist.rvs( size )

integral = (max(data)-min(data)) * np.mean(func(data))
print(integral)

```
